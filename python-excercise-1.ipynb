### Kasus 6.2 — Fungsi Hitung Luas Persegi Panjang

### Kasus 6.3 — Fungsi dengan Nilai Default

**Skenario:** Toko online punya diskon standar 10% untuk semua pelanggan, tapi kadang ada diskon khusus untuk member VIP.

**Pseudocode:**
```

    kembalikan harga - (harga * persen/100)

cetak hitung_diskon(100000)
cetak hitung_diskon(100000, 25)
```

### Kasus 6.4 — Fungsi Cek Bilangan Prima

**Skenario:** Aplikasi edukasi matematika ingin memberi tahu siswa apakah angka yang mereka masukkan termasuk bilangan prima atau bukan.

**Pseudocode:**
```
fungsi cek_prima(angka):
    jika angka kurang dari 2:
        kembalikan salah
    untuk i dari 2 sampai angka-1:
        jika angka habis dibagi i:
            kembalikan salah
    kembalikan benar
```

### Kasus 6.5 — Fungsi Rata-rata Nilai

**Skenario:** Guru ingin fungsi yang bisa menerima daftar nilai siswa (list) dan langsung mengembalikan nilai rata-ratanya.

**Pseudocode:**
```
fungsi rata_rata(daftar_nilai):
    total = jumlahkan semua nilai
    kembalikan total / banyaknya nilai

cetak rata_rata([80, 90, 70, 85])
```

### Kasus 6.6 — Fungsi Validasi Password

**Skenario:** Sistem pendaftaran akun ingin memvalidasi apakah password yang dimasukkan pengguna memenuhi syarat minimal 8 karakter.

**Pseudocode:**
```
fungsi validasi_password(password):
    jika panjang(password) >= 8:
        kembalikan "valid"
    kembalikan "terlalu pendek"
```

### Kasus 6.7 — Fungsi dengan Banyak Nilai Kembali

**Skenario:** Aplikasi olahraga ingin fungsi yang menghitung sekaligus mengembalikan nilai maksimum DAN minimum dari daftar kecepatan lari seorang atlet.

**Pseudocode:**
```
fungsi min_max(daftar_kecepatan):
    cari nilai_min
    cari nilai_max
    kembalikan nilai_min, nilai_max

min_val, max_val = min_max([12, 15, 9, 20, 11])
```

### Kasus 6.8 — Fungsi Rekursif Sederhana (Faktorial)

**Skenario:** Untuk memperdalam logika fungsi, buat fungsi yang menghitung faktorial suatu angka menggunakan pemanggilan fungsi itu sendiri (rekursi).

**Pseudocode:**
```
fungsi faktorial(n):
    jika n == 0 atau n == 1:
        kembalikan 1
    kembalikan n * faktorial(n - 1)
```
